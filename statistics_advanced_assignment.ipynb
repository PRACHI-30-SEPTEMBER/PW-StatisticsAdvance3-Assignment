## Q1: What is Estimation Statistics?

**Answer:**
Estimation statistics involves using sample data to estimate population parameters.

- Point Estimate: Single value (e.g., sample mean)
- Interval Estimate: Range of values (confidence interval)

## Q2: Python function to estimate population mean

**Answer:**

```
def estimate_mean(sample):
    return sum(sample)/len(sample)

data = [10, 20, 30, 40, 50]
estimate_mean(data)
```

## Q3: What is Hypothesis Testing?

**Answer:**
It is a statistical method to test assumptions about population.

Importance:
- Decision making
- Validating claims

## Q4: Hypothesis (Male vs Female weight)

**Answer:**
H0: Mean weight of males ≤ females
H1: Mean weight of males > females

## Q5: Python script for two-sample test

**Answer:**

```
import scipy.stats as stats

sample1 = [10,12,14,16,18]
sample2 = [8,9,11,13,15]

t_stat, p_val = stats.ttest_ind(sample1, sample2)
print(t_stat, p_val)
```

## Q6: Null & Alternative Hypothesis

## Q8: p-value

**Answer:**
p-value is probability of observing results under H0.
Lower p-value → reject H0

## Q9: t-distribution plot

**Answer:**

```
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t

x = np.linspace(-5,5,100)
y = t.pdf(x, df=10)

plt.plot(x,y)
plt.title("t-distribution (df=10)")
plt.show()
```

## Q10: Two-sample t-test program

**Answer:**

```
import scipy.stats as stats

a = [10,12,14,16]
b = [9,11,13,15]

t,p = stats.ttest_ind(a,b)
print("t:",t,"p:",p)
```
